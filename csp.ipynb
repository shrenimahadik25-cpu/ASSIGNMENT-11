```python
# Constraint Satisfaction Problem (CSP)
# Map Coloring using Backtracking and Constraint Propagation

# Step 1: Define the variables (regions)
regions = ['WA', 'NT', 'SA', 'Q', 'NSW', 'V', 'T']

# Step 2: Define the domain (available colors)
colors = ['Red', 'Green', 'Blue']

# Step 3: Define the constraints (neighboring regions)
neighbors = {
    'WA': ['NT', 'SA'],
    'NT': ['WA', 'SA', 'Q'],
    'SA': ['WA', 'NT', 'Q', 'NSW', 'V'],
    'Q': ['NT', 'SA', 'NSW'],
    'NSW': ['Q', 'SA', 'V'],
    'V': ['SA', 'NSW'],
    'T': []
}

# Step 4: Check whether a color is valid
def is_consistent(region, color, assignment):
    for neighbor in neighbors[region]:
        if neighbor in assignment and assignment[neighbor] == color:
            return False
    return True


# Step 5: Select an unassigned region using MRV
# MRV = Minimum Remaining Values
def select_unassigned_region(assignment, domains):
    unassigned = [r for r in regions if r not in assignment]

    return min(unassigned, key=lambda r: len(domains[r]))


# Step 6: Backtracking with constraint propagation
def backtracking(assignment, domains):

    # If every region has a color, return the solution
    if len(assignment) == len(regions):
        return assignment

    # Select the region with the fewest available colors
    region = select_unassigned_region(assignment, domains)

    # Try each available color
    for color in domains[region]:

        if is_consistent(region, color, assignment):

            # Assign the color
            assignment[region] = color

            # Save the current domains
            new_domains = {
                r: list(domains[r]) for r in regions
            }

            # Constraint propagation:
            # Remove this color from neighboring regions
            valid = True

            for neighbor in neighbors[region]:

                if neighbor not in assignment:
                    if color in new_domains[neighbor]:
                        new_domains[neighbor].remove(color)

                    # If no colors remain, this choice fails
                    if len(new_domains[neighbor]) == 0:
                        valid = False
                        break

            # Continue searching if propagation succeeds
            if valid:
                result = backtracking(assignment, new_domains)

                if result is not None:
                    return result

            # Backtrack: remove the assignment
            del assignment[region]

    # No solution found for this branch
    return None


# Step 7: Initialize domains
domains = {
    region: colors.copy() for region in regions
}

# Step 8: Solve the CSP
solution = backtracking({}, domains)

# Step 9: Display the result
print("CSP - MAP COLORING SOLUTION")
print("-" * 35)

if solution:
    for region in regions:
        print(region, ":", solution[region])

    print("\nSolution found successfully!")

    # Verify the constraints
    valid = all(
        solution[region] != solution[neighbor]
        for region in regions
        for neighbor in neighbors[region]
    )

    print("All constraints satisfied:", valid)

else:
    print("No solution exists.")
```
